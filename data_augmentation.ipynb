```python
import os
from tensorflow.keras.preprocessing.image import ImageDataGenerator, img_to_array, load_img
```

```python
# =========================
# BASE DIRECTORY
# =========================
base_dir = "/Users/theabood/Desktop/DataScience/semester3/Datasets/products"

# =========================
# DATA AUGMENTATION CONFIG
# =========================
datagen = ImageDataGenerator(
    rotation_range=40,
    width_shift_range=0.2,
    height_shift_range=0.2,
    shear_range=0.2,
    zoom_range=0.2,
    horizontal_flip=True,
    fill_mode='nearest'
)

# =========================
# LOOP THROUGH PRODUCT FOLDERS
# =========================
for product_name in os.listdir(base_dir):

    source_dir = os.path.join(base_dir, product_name)

    # Skip non-folder files
    if not os.path.isdir(source_dir) or product_name == "train":
        continue

    print(f"Processing product: {product_name}")

    # Create train directory for each product
    train_dir = os.path.join(base_dir, "train", product_name)
    os.makedirs(train_dir, exist_ok=True)

    # =========================
    # LOOP THROUGH IMAGES
    # =========================
    for filename in os.listdir(source_dir):

        if filename.lower().endswith(('.png', '.jpg', '.jpeg')):

            img_path = os.path.join(source_dir, filename)

            img = load_img(img_path)
            x = img_to_array(img)
            x = x.reshape((1,) + x.shape)

            # Generate augmented images
            i = 0
            for batch in datagen.flow(
                x,
                batch_size=1,
                save_to_dir=train_dir,
                save_prefix=product_name,
                save_format='jpg'
            ):
                i += 1
                if i > 30:   # 30 augmentations per image
                    break
```

```python
import os
import random
import shutil

# =========================
# BASE DIRECTORY
# =========================
base_dir = "/Users/theabood/Desktop/DataScience/semester3/products"

train_dir = os.path.join(base_dir, "train")
test_dir = os.path.join(base_dir, "test")

# Create test directory if it doesn't exist
os.makedirs(test_dir, exist_ok=True)

# =========================
# LOOP THROUGH PRODUCT CLASSES
# =========================
for product_name in os.listdir(train_dir):

    product_train_path = os.path.join(train_dir, product_name)

    # Skip non-directories
    if not os.path.isdir(product_train_path):
        continue

    print(f"Processing product: {product_name}")

    # Create corresponding test folder
    product_test_path = os.path.join(test_dir, product_name)
    os.makedirs(product_test_path, exist_ok=True)

    # Get all image files
    images = [
        f for f in os.listdir(product_train_path)
        if f.lower().endswith(('.png', '.jpg', '.jpeg'))
    ]

    # Shuffle images randomly
    random.shuffle(images)

    # Calculate 20%
    split_count = int(0.2 * len(images))

    # Select 20% images
    test_images = images[:split_count]

    # Move images
    for img in test_images:
        src = os.path.join(product_train_path, img)
        dst = os.path.join(product_test_path, img)
        shutil.move(src, dst)

print("✅ 20% of images moved from train to test for each product class.")

```

```python
import os
import random
import shutil

base_dir = "/Users/theabood/Desktop/DataScience/semester3/Fruits"

train_dir = os.path.join(base_dir, "train")
test_dir = os.path.join(base_dir, "test")

# Create test directory if it doesn't exist
os.makedirs(test_dir, exist_ok=True)

# Loop through each fruit class folder
for fruit_name in os.listdir(train_dir):
    
    fruit_train_path = os.path.join(train_dir, fruit_name)
    
    # Skip if not a directory
    if not os.path.isdir(fruit_train_path):
        continue

    # Create corresponding test subfolder
    fruit_test_path = os.path.join(test_dir, fruit_name)
    os.makedirs(fruit_test_path, exist_ok=True)

    # List all image files
    images = [f for f in os.listdir(fruit_train_path)
              if f.lower().endswith(('.png', '.jpg', '.jpeg'))]

    # Shuffle randomly
    random.shuffle(images)

    # Calculate 20%
    split_count = int(0.2 * len(images))

    # Select 20% images
    test_images = images[:split_count]

    # Move images to test folder
    for img in test_images:
        src = os.path.join(fruit_train_path, img)
        dst = os.path.join(fruit_test_path, img)
        shutil.move(src, dst)

print("✅ 20% of images moved from train to test for each product class.")
```
