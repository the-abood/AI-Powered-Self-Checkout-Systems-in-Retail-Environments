```python
import torch
import os

# Load model from Google Drive
model = torch.hub.load('ultralytics/yolov5', 'custom',
                       path='/content/drive/MyDrive/yolov5_results/weights/best.pt')

# Get all JPG images from folder
image_folder = '/content/drive/MyDrive/test_images/'
images = [os.path.join(image_folder, f) for f in os.listdir(image_folder) if f.endswith('.jpg')]

print(f'Found {len(images)} images')

# Run detection
results = model(images)

# Save results back to Google Drive
results.print()
results.save(save_dir='/content/drive/MyDrive/detection_results/')
```
